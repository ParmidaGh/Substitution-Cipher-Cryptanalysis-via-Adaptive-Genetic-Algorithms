## Using Adaptive Mutation Rate

```python
import string
import random
import re
import numpy as np

population_Qty = 50
crossover_rate = 0.7
truncation_rate = 0.3

class Decoder:

    def __init__(self, encoded):
        self.encoded = encoded
        self.encoded_words = self.clean_dictionary(self.encoded.lower(), is_encoded=True)
        self.dictionary = self.clean_dictionary(open("/content/drive/MyDrive/Data/global_text.txt").read().lower())
        self.mutation_rate = 0.1  # Starting mutation rate
        self.iter_without_improvement = 0
        self.max_iter_without_improvement = 10

    def clean_dictionary(self, dictionary, is_encoded=False):
        dictionary = re.sub(r'[^A-Za-z]', ' ', dictionary)
        words = dictionary.split()
        words = list(dict.fromkeys(words))
        if is_encoded:
            return words
        else:
            return set(words)

    def generate_rand_chromosome(self):
        chromosomes = set(string.ascii_lowercase)
        new_chromosome = {}
        for c in string.ascii_lowercase:
            chromosome = random.choice(list(chromosomes))
            chromosomes.remove(chromosome)
            new_chromosome[c] = chromosome
        return new_chromosome

    def decode_with_key(self, key, text=None):
        if text is None:
            text = self.encoded
        if isinstance(text, list):
            return [self.decode_with_key(key, word) for word in text]
        decoded_text = ''
        for char in text:
            if not char.islower():
                decoded_text += key.get(char.lower(), char).upper()
            else:
                decoded_text += key.get(char, char)
        return decoded_text

    def crossover(self, parent1, parent2): # one-point crossover
        parent1_list = list(parent1.values())
        parent2_list = list(parent2.values())
        size = len(parent1_list)
        point = random.randint(1, size - 1)
        tmp = parent1_list[:point]
        parent2_list = [x for x in parent2_list if x not in tmp]
        newChromosome = tmp + parent2_list

        # make new chromosome
        new_char = {}
        for c in string.ascii_lowercase:
            char = newChromosome[0]
            newChromosome.remove(char)
            new_char[c] = char
        return new_char

    def fitness_calculator(self, population):
        scores = []
        for i in range(len(population)):
            chromosome = population[i]
            decoded_words = self.decode_with_key(chromosome, self.encoded_words)
            counter = 0
            for word in decoded_words:
                if word in self.dictionary:
                    counter += 1
            scores.append((counter, i))
        scores.sort(key=lambda score: score[0], reverse=True)
        return scores

    def mutation(self, chromosome): # inversion mutation
        size = len(list(chromosome.values()))
        point1 = random.randint(0, size - 1)
        point2 = random.randint(0, size - 1)

        if point2 >= point1:
            point2 += 1
        else:
            point1, point2 = point2, point1

        vals = list(chromosome.values())
        keys = list(chromosome.keys())
        segment = vals[point1:point2]
        segment.reverse()
        vals[point1:point2] = segment
        chromosome = dict(zip(keys, vals))
        return chromosome

    def decode(self):

        # genetic algorithm
        population = [self.generate_rand_chromosome() for i in range(population_Qty)]

        previous_best_score = 0

        while True:
            scores = []
            for i in range(len(population)):
                chromosome = population[i]
                decoded_words = self.decode_with_key(chromosome, self.encoded_words)
                counter = 0
                for word in decoded_words:
                    if word in self.dictionary:
                        counter += 1
                scores.append((counter, i))
            scores.sort(key=lambda score: score[0], reverse=True)

            if scores[0][0] >= len(self.encoded_words):
                best_chromosome_index = scores[0][1]
                key = population[best_chromosome_index]
                print("Chromosome (final key mapping):\n ", key)
                print("\nDecoded text is:\n ")
                return self.decode_with_key(key)

            # update adaptive mutation rate based on performance
            best_score = scores[0][0]
            if best_score > previous_best_score:
                self.iter_without_improvement = 0
            else:
                self.iter_without_improvement += 1
                if self.iter_without_improvement >= self.max_iter_without_improvement:
                    self.mutation_rate *= 2  # doubling mutation rate for exploration
                    self.iter_without_improvement = 0

            previous_best_score = best_score

            # Truncation selection
            new_population = [population[i] for score, i in scores[:int(population_Qty * truncation_rate)]]
            size = int(crossover_rate * population_Qty)
            for _ in range(size):
                parents = np.random.choice(new_population, 2, replace=False)
                child = self.crossover(parents[0], parents[1])
                new_population.append(self.mutation(child))
            population = new_population

encoded = open("/content/drive/MyDrive/Data/encoded_text.txt").read()
d = Decoder(encoded)
decoded_text = d.decode()
print(decoded_text)
```
